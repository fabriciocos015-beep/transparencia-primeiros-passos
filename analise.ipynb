```python
import pandas as pd

df = pd.read_csv("202608_Despesas.csv", sep=";", encoding="latin-1")
df.head()
```

```python
df.columns.tolist()
```

```python
#converted Brazilian format to float
df['Valor Pago (R$)']= (
df['Valor Pago (R$)'].str.replace('.','', regex=False).str.replace(',','.', regex=False).astype(float)
)
df.head()
```

```python
#Question 1 — Top 10 órgãos by amount paid:

df.groupby('Nome Órgão Superior')['Valor Pago (R$)'].sum().sort_values(ascending=False).head(10)
```

```python
#Question 2 — Total paid per expense group:

df.groupby('Nome Grupo de Despesa')['Valor Pago (R$)'].sum().sort_values(ascending=False)
```

```python
# Question 3 — Duplicate rows:

df.duplicated().sum()
```
